```python
import numpy as np
from sklearn.impute import SimpleImputer

# 1. Create a 3x3 matrix with missing values (np.nan)
matrix_with_nans = np.array([
    [1,     2 , np.nan],
    [4,      np.nan,      6],
    [np.nan, 8,       9]
])

print("Original 3x3 Matrix with NaNs:")
print(matrix_with_nans)

# 2. Initialize the SimpleImputer with the univariate 'mean' strategy
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')

# 3. Fit the imputer and transform the matrix
imputed_matrix = imputer.fit_transform(matrix_with_nans)

print("\nImputed Matrix (Missing values replaced by column means):")
print(imputed_matrix)
```
